```python
import glob
import os

directory_path = '/content/drive/MyDrive/Colab Notebooks/Research/Emotiv 30s EDF'
search_path = os.path.join(directory_path, '**/*.edf')
edf_files = sorted(glob.glob(search_path, recursive=True))

print(f"Found {len(edf_files)} EDF files.")
```

```python
import mne
filtered_signals = []
print(f"Re-applying band-pass filtering for {len(edf_files)} files...")
for i, file_path in enumerate(edf_files):
    temp_raw = mne.io.read_raw_edf(file_path, preload=True, verbose=False)
    temp_raw.filter(l_freq=1, h_freq=40.0, verbose=False)
    filtered_signals.append(temp_raw)
    if (i + 1) % 20 == 0: print(f"Filtered {i + 1}/{len(edf_files)} files...")
```

```python
# Define a function to apply a notch filter
def apply_notch_filter(raw_list, notch_freq=50.0):
    notched_signals = []
    print(f"Starting notch filtering at {notch_freq} Hz for {len(raw_list)} files...")
    for i, raw_temp in enumerate(raw_list):
        # Apply the notch filter
        raw_temp.notch_filter(freqs=notch_freq, verbose=False)
        notched_signals.append(raw_temp)

        if (i + 1) % 10 == 0:
            print(f"Processed {i + 1}/{len(raw_list)} files...")

    print("\nFinished applying notch filter to all files.")
    return notched_signals

# Apply the notch filter to the previously band-pass filtered signals
cleaned_signals_notch = apply_notch_filter(filtered_signals)

print(f"Total signals after notch filtering: {len(cleaned_signals_notch)}")
```

```python
print(f"Starting Common Average Referencing (CAR) for {len(filtered_signals)} files (applied to filtered_signals)...")

# Create a list to store the re-referenced raw objects from filtered_signals
filtered_signals_car = []

for i, raw_obj in enumerate(filtered_signals):
    # Create a copy to avoid modifying the original raw object
    raw_car = raw_obj.copy()

    # Set the reference to 'average'. This will apply CAR.
    # It finds all EEG channels and sets them to the average reference.
    raw_car.set_eeg_reference('average', verbose=False)
    filtered_signals_car.append(raw_car)

    if (i + 1) % 10 == 0:
        print(f"Re-referencing processed for {i + 1}/{len(filtered_signals)} files...")

print("\nFinished applying Common Average Referencing to filtered_signals.")
print(f"Total signals after CAR on filtered_signals: {len(filtered_signals_car)}")
```

```python
from mne.preprocessing import ICA
cleaned_all_signals = []
exclude_idx = [0, 1]
print("Restoring cleaned_all_signals via ICA...")
for i, raw_notched in enumerate(filtered_signals):
    raw_notched.notch_filter(freqs=50.0, verbose=False)
    ica = ICA(n_components=0.999, max_iter='auto', random_state=97)
    ica.fit(raw_notched, verbose=False)
    raw_cleaned = raw_notched.copy()
    ica.exclude = exclude_idx
    ica.apply(raw_cleaned, verbose=False)
    cleaned_all_signals.append(raw_cleaned)
    if (i + 1) % 20 == 0: print(f"ICA complete for {i + 1} files...")
```

```python
print(f"Total cleaned signals after ICA: {len(cleaned_all_signals)}")

# Display the first cleaned signal after ICA
if cleaned_all_signals:
    print("Displaying the first cleaned signal after ICA...")
    cleaned_all_signals[0].plot(block=True, show=True)
else:
    print("No cleaned signals available to display.")
```

```python
import numpy as np

nan_found = False
inf_found = False

for i, raw in enumerate(cleaned_all_signals):
    data = raw.get_data()
    if np.isnan(data).any():
        print(f"NaN detected in signal index {i}")
        nan_found = True
    if np.isinf(data).any():
        print(f"Inf detected in signal index {i}")
        inf_found = True

if not nan_found and not inf_found:
    print("Data check passed: No NaN or Inf values found in the cleaned signals.")
else:
    print("Data check failed: NaN or Inf values were detected.")
```

```python
from mne.preprocessing import ICA, annotate_amplitude
cleaned_all_signals = []
exclude_idx = [0, 1]
print("Processing ICA and final annotations...")
for i, raw_filt in enumerate(filtered_signals):
    raw_filt.notch_filter(freqs=50.0, verbose=False)
    ica = ICA(n_components=0.999, max_iter='auto', random_state=97)
    ica.fit(raw_filt, verbose=False)
    raw_cleaned = raw_filt.copy()
    ica.exclude = exclude_idx
    ica.apply(raw_cleaned, verbose=False)
    cleaned_all_signals.append(raw_cleaned)

reject_criteria = dict(eeg=100e-6)
final_cleaned_signals = []
for raw in cleaned_all_signals:
    raw_final = raw.copy()
    bad_annots, _ = annotate_amplitude(raw_final, peak=reject_criteria, verbose=False)
    raw_final.set_annotations(raw_final.annotations + bad_annots)
    final_cleaned_signals.append(raw_final)
print(f"Done! {len(final_cleaned_signals)} signals ready.")
```

```python
# Extract data from the first processed signal to calculate statistics
signal = final_cleaned_signals[0].get_data()

print("Signal shape (channels x samples):", signal.shape)
print("Standard deviation (all channels):", np.std(signal))
print("Minimum value:", np.min(signal))
print("Maximum value:", np.max(signal))
```

```python
import pandas as pd
import numpy as np

# Assuming final_cleaned_signals is a list of MNE Raw objects
# Get sampling frequency from the first signal
sfreq = final_cleaned_signals[0].info['sfreq']

segment_duration_seconds = 60
samples_per_segment = int(sfreq * segment_duration_seconds)

segmented_data_list = []

print(f"Segmenting {len(final_cleaned_signals)} files into 60-second intervals...")

for i, raw in enumerate(final_cleaned_signals):
    data = raw.get_data() # channels x samples
    file_name = edf_files[i].split('/')[-1]

    # Ensure the signal is at least 3 * segment_duration_seconds long (180 seconds)
    if raw.n_times < 3 * samples_per_segment:
        print(f"Skipping file {file_name}: signal too short ({raw.n_times / sfreq:.1f}s) for 3 segments.")
        continue

    # Calculate start and end samples for each segment
    start_low_1 = 0
    end_low_1 = samples_per_segment

    start_high = samples_per_segment
    end_high = 2 * samples_per_segment

    start_low_2 = raw.n_times - samples_per_segment
    end_low_2 = raw.n_times

    for ch_idx, ch_name in enumerate(raw.ch_names):
        # Segment 1: First 60 seconds (Low Concentration)
        segment_1 = data[ch_idx, start_low_1:end_low_1]

        # Segment 3: Last 60 seconds (Low Concentration)
        segment_3 = data[ch_idx, start_low_2:end_low_2]

        # Combine first and last low concentration segments
        combined_low_concentration_segment = np.concatenate((segment_1, segment_3))

        segmented_data_list.append({
            'File': file_name,
            'Channel': ch_name,
            'Segment_Type': 'Low Concentration',
            'Segment_Index': 1,
            'Data': combined_low_concentration_segment,
            'Duration_s': 2 * segment_duration_seconds # Duration is now 120 seconds
        })

        # Segment 2: Middle 60 seconds (High Concentration)
        segment_2 = data[ch_idx, start_high:end_high]
        segmented_data_list.append({
            'File': file_name,
            'Channel': ch_name,
            'Segment_Type': 'High Concentration',
            'Segment_Index': 2,
            'Data': segment_2,
            'Duration_s': segment_duration_seconds
        })

    if (i + 1) % 20 == 0:
        print(f"Processed {i + 1}/{len(final_cleaned_signals)} files...")

segmented_df = pd.DataFrame(segmented_data_list)

print("\nSegmentation complete.")
print(f"Total segments created: {len(segmented_df)}")
display(segmented_df.head())
```

```python
import pyeeg
from itertools import chain
import numpy as np

def extract_features(signal):
    aa = pyeeg.bin_power(signal, [0.5, 4, 7, 12, 30], 173)
    power_list = list(aa[0])
    ratio_list = list(aa[1])
    b = [
        [pyeeg.hfd(signal, Kmax=5)],
        [pyeeg.svd_entropy(signal, Tau=4, DE=10)],
        [pyeeg.fisher_info(signal, Tau=4, DE=10)],
        [pyeeg.dfa(signal)]
    ]
    flattened_b = list(chain.from_iterable(b))
    return power_list + ratio_list + flattened_b

if 'final_cleaned_signals' in globals() and len(final_cleaned_signals) > 0:
    test_signal = final_cleaned_signals[0].get_data()[0, :]
    test_features = extract_features(test_signal)
    print("Test Success! Features:", test_features)
else:
    print("Warning: Data not found. Please ensure cells above ran successfully.")
```

```python
import pandas as pd

all_features_list = []

print(f"Extracting features from {len(segmented_df)} segments...")

for i, row in enumerate(segmented_df.itertuples(index=False)):
    # Access data and metadata directly from the DataFrame row
    signal_data = row.Data
    file_name = row.File
    channel_name = row.Channel
    segment_type = row.Segment_Type

    try:
        feat_vector = extract_features(signal_data)

        all_features_list.append({
            'File': file_name,
            'Channel': channel_name,
            'Segment_Type': segment_type, # Add Segment_Type
            'Delta_P': feat_vector[0],
            'Theta_P': feat_vector[1],
            'Alpha_P': feat_vector[2],
            'Beta_P': feat_vector[3],
            'Delta_R': feat_vector[4],
            'Theta_R': feat_vector[5],
            'Alpha_R': feat_vector[6],
            'Beta_R': feat_vector[7],
            'HFD': feat_vector[8],
            'SVD_Entropy': feat_vector[9],
            'Fisher_Info': feat_vector[10],
            'DFA': feat_vector[11]
        })
    except Exception as e:
        print(f"Error extracting features for {file_name}, channel {channel_name}, segment type {segment_type}: {e}")
        continue

    if (i + 1) % 500 == 0: # Adjusted print frequency for segments
        print(f"Processed {i+1}/{len(segmented_df)} segments...")

final_df = pd.DataFrame(all_features_list)
final_df.to_csv('pyeeg_processed_features.csv', index=False)

print('\nFeature extraction complete.')
print(f"Saved results to 'pyeeg_processed_features.csv'. Total rows: {len(final_df)}")
display(final_df.head())
```

```python
print("Original class distribution:")
print(final_df['Segment_Type'].value_counts())

# Separate data into concentration types
high_concentration_df = final_df[final_df['Segment_Type'] == 'High Concentration']
low_concentration_df = final_df[final_df['Segment_Type'] == 'Low Concentration']

# Find the minimum number of samples between the two classes
min_samples = min(len(high_concentration_df), len(low_concentration_df))

# Downsample the larger class to match the smaller class
balanced_high_concentration_df = high_concentration_df.sample(n=min_samples, random_state=42)
balanced_low_concentration_df = low_concentration_df.sample(n=min_samples, random_state=42)

# Concatenate the balanced dataframes
balanced_df = pd.concat([balanced_high_concentration_df, balanced_low_concentration_df])

# Shuffle the balanced dataframe to mix the classes
balanced_df = balanced_df.sample(frac=1, random_state=42).reset_index(drop=True)

print("\nBalanced dataset created.")
print("New class distribution:")
print(balanced_df['Segment_Type'].value_counts())
display(balanced_df.head())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Define features (X) and target (y)
X = balanced_df.drop(['File', 'Channel', 'Segment_Type'], axis=1)
y = balanced_df['Segment_Type']

# Split the dataset into training and testing sets (80% train, 20% test)
# Stratify by 'Segment_Type' to ensure class distribution is maintained
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Reconstruct the train and test dataframes for easier handling if needed
train_df = pd.concat([X_train, y_train], axis=1)
test_df = pd.concat([X_test, y_test], axis=1)

print("Dataset split into training and testing sets.")
print(f"Training set shape: {train_df.shape}")
print(f"Testing set shape: {test_df.shape}")

print("\nTraining set class distribution:")
print(train_df['Segment_Type'].value_counts())

print("\nTesting set class distribution:")
print(test_df['Segment_Type'].value_counts())

display(train_df.head())
display(test_df.head())
```

```python
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Feature Scaling
# SVMs are sensitive to the scale of the features, so it's good practice to scale them.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Features scaled successfully.")

# 2. Initialize and Train the SVM Classifier
# Using a Support Vector Classifier (SVC) with a radial basis function (RBF) kernel as a common choice.
svm_model = SVC(kernel='rbf', random_state=42)
print("\nTraining SVM model...")
svm_model.fit(X_train_scaled, y_train)
print("SVM model training complete.")

# 3. Make Predictions
y_pred = svm_model.predict(X_test_scaled)

# 4. Evaluate the Model
print("\n--- SVM Model Evaluation ---")

# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")

# Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Confusion Matrix
print("\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred)
print(cm)

# Plotting Confusion Matrix for better visualization
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=svm_model.classes_, yticklabels=svm_model.classes_)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for SVM')
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Initialize and Train the Random Forest Classifier
# Using n_estimators=100 (default) and random_state for reproducibility
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
print("\nTraining Random Forest model...")
rf_model.fit(X_train_scaled, y_train)
print("Random Forest model training complete.")

# Make Predictions
y_pred_rf = rf_model.predict(X_test_scaled)

# Evaluate the Model
print("\n--- Random Forest Model Evaluation ---")

# Accuracy
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print(f"Accuracy: {accuracy_rf:.4f}")

# Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))

# Confusion Matrix
print("\nConfusion Matrix:")
cm_rf = confusion_matrix(y_test, y_pred_rf)
print(cm_rf)

# Plotting Confusion Matrix for better visualization
plt.figure(figsize=(6, 5))
sns.heatmap(cm_rf, annot=True, fmt='d', cmap='Greens',
            xticklabels=rf_model.classes_, yticklabels=rf_model.classes_)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix for Random Forest')
plt.show()
```
